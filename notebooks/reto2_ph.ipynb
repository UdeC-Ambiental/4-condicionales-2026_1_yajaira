# 🧩 Reto 2 – pH
**Objetivo:** clasificar pH y generar alerta química.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `pH`.
3. Crea `clasificacion_pH` con tres rangos.
4. Crea `alerta_quimica` (revisión si pH < 6.5 o pH > 8.5).
5. Filtra `df_alerta` y exporta `resultado_ph.csv`.

> **Reto adicional:** promedio y desviación estándar.

```python
1# Carga el CSV en df
import pandas as pd


df = pd.read_csv("../data/muestras_calidad_agua.csv")

df.info()
```

```python
2# Extraer sitio y pH df_pH

df_pH = df[["sitio", "pH"]]

df_pH
```

```python
3# Crea clasificacion_pH con tres rangos.

import numpy as np

df_pH["clasificacion_pH"] = np.where(
    df_pH["pH"] < 6.5,
    "Ácido",
    np.where(
        df_pH["pH"] <= 8.5,
        "Neutro",
        "Básico"
    )
)

df_pH
```

```python
4# Crea alerta_quimica (revisión si pH < 6.5 o pH > 8.5)
def alerta(ph):
    if ph < 6.5 or ph > 8.5:
        return "Sí"
    else:
        return "No"

df_pH["alerta_quimica"] = df_pH["pH"].apply(alerta)

df_pH
```

```python
5# Filtra df_alerta y exporta resultado_ph.csv
```

```python
df_alerta = df_pH[df_pH["alerta_quimica"] == "Sí"]

df_alerta
```

```python
df_alerta.to_csv("resultado_ph.csv", index=False)
```

```python
##RETO ADICIONAL
#Reto adicional:promedio y desviación estándar.
promedio = df_pH["pH"].mean()

print("Promedio de pH:", promedio)


desviacion = df_pH["pH"].std()

print("Desviación estándar:", desviacion)
```
